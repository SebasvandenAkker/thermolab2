# Bepaling van soortelijke warmte van water


# Introductie

Water heeft een enorme buffercapaciteit: je moet veel energie toevoegen om de temperatuur van water een graad te verwarmen. In dit practicum gaan we de soortelijke warmte van water bepalen door een bekende hoeveelheid water te verwarmen met een bekende hoeveelheid energie, en de temperatuurstijging te meten.

# Theorie

De benodigde energie die nodig is om water een bepaalde hoeveelheid kelvin op te laten warmen kan worden gegeven door de formule 
$$
E = mc_w\Delta T
$$
Hierin is $E$ de hoeveelheid geleverde energie in Joule, $m$ de massa van het water, $\Delta T$ de hoeveelheid kelvin dat het water is toegenomen en $c_w$ de soortelijke warmte van water.

## Data analyse
Wat we doen is we fitten onze data over de vorm $y= ax + b$ met de temperatuur als y variabel en de t als x. Dit is gebaseerd op de formule $E = mC\Delta T + T_0$ deze kan je omschrijven naar $T = \frac{P}{mc}t + T_0$ en dus van de vorm waarover we fitten. Uit deze fit krijgen we dan een a, gelijk aan $\frac{P}{mc}$ hieruit kunnen we dan makkelijk c berekenen.

# Resultaten


````python
# Hier de data en de analyse
import numpy as np
from matplotlib import pyplot as plt
from scipy.optimize import curve_fit

# Hieronder staan onze meetwaardes;
tijd = np.linspace(0, 480,  13) # De tijden waarop wij gemeten hebben secondes
temp = np.array([27.1, 27.7, 28.2, 28.8, 29.4, 30.0, 30.5, 31.1, 31.7, 32.3, 32.9, 33.5, 34.1]) # De temperaturen die wij op die tijdstippen hebben gemeten in graden celcius
U = 17.1 # gebruikte spanning in volt
I = 1.76 # de bijborende stroomspanning in Ampere
P = U * I # Het vermogen van de weerstand in J/s
m = 0.4845 # De massa van het water

# Hieronder is de code voor het functiefitten
def lineair(t, a, b): # de functie die we gebruiken voor het functiefitten
    return a * t + b

params, cov = curve_fit(lineair, tijd, temp) # de functiefit
a, b = params # de berekende helling
helling = a  # K/s
print(f"De berekende soortelijke warmte is {P/(a*m)} J/(kg * K)") # J/kg K

# Hieronder is de code voor het plotten van de meetpunten en de functiefit
t_fit = np.linspace(0, 480, 101)
plt.plot(tijd, temp, 'o', label='Metingen')
plt.plot(t_fit, lineair(t_fit, a, b), 'r--', label='Trendlijn (fit)')
plt.xlabel('Tijd (s)')
plt.ylabel('Temperatuur (°C)')
plt.title("T/t-diagram")
plt.legend()
plt.show()
plt.savefig("naam.png", dpi=450)
````

# Discussie en conclusie

Onze berekende soortelijke warmte van water is 4266.19 J/(kg * K). Dit komt eenigsins in de beurt van de theoretische waarde van 4186 J/(kg * K), maar zit er wat boven. Het feit dat onze gevonden waarde hoger is dan de daadwerkelijke waarde is echter wel logisch. Wij hebben namelijk een aannamen gemaakt dat het vermogen dat de weerstand leverd voor 100% wordt omgezet in warmte in het water, in realiteit zal dit iets lager dan 100% zijn. Daarnaast hebben wij geen rekening gehouden met het verlies van warmte naar de buitenlucht, in realiteit wordt tijdens het opwarmen van het water wat energie weggegeven aan de omgeving. Deze twee feiten zorgen ervoor dat wij meer tijd moeten spenderen om het water naar een bepaalde temperatuur te krijgen dan we in theorie verwachten. Hierdoor is onze gevonden soortelijke waarde hoger dan de theoretische waarde.
